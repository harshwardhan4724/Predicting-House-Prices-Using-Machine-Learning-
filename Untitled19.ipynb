```python
%%bash

echo "Starting target application..."

export PORT=3000

# Start Node.js app
if [ -f package.json ]; then
  if jq -e '.scripts.start' package.json >/dev/null 2>&1; then
    echo "Running npm start..."
    nohup npm start >/tmp/app.log 2>&1 &
    sleep 8
  fi
fi

# Start Spring Boot app
if [ -f pom.xml ]; then
  echo "Running mvn spring-boot:run..."
  nohup mvn -q -DskipTests spring-boot:run >/tmp/mvn-run.log 2>&1 &
  sleep 10
fi

echo "Checking if target URL is reachable: ${TARGET_URL}"
for i in {1..30}; do
  if curl -sSf "${TARGET_URL}" >/dev/null 2>&1; then
    echo "Application is up!"
    break
  else
    echo "Waiting for app... ($i)"
    sleep 1
  fi
done
```
